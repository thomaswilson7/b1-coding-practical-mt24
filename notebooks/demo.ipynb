# Simulate closed-loop
After implementing your control functionality, you can simulate the closed-loop with code that looks something like this...

```python
# Import relevant modules
import sys
from pathlib import Path

# Makes it so that you can import modules from the parent directory of the current working directory
current_dir = Path.cwd()
repo_dir = current_dir.parent
sys.path.append(str(repo_dir))

# Imports classes
from uuv_mission.dynamic import ClosedLoop, Submarine, Mission
from uuv_mission.control import Controller

# Initialises controller and submarine
controller = Controller()
sub = Submarine()

# Creates closed loop system
closed_loop = ClosedLoop(sub, controller)

# Imports mission data
mission = Mission.from_csv(str(repo_dir/"data"/"mission.csv"))

# Simulate the mission with random disturbances
trajectory = closed_loop.simulate_with_random_disturbances(mission)

# Plot the trajectory of the completed mission
trajectory.plot_completed_mission(mission)
```
